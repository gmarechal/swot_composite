```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
path_input = '../data/'
path_output = '../figures/'
```

```python
ds_GS = xr.open_dataset(path_input + 'GS2_data_figS7.nc')
ds_AGULHAS = xr.open_dataset(path_input + 'AGULHAS_data_figS7.nc')
ds_CCS = xr.open_dataset(path_input + 'CCS_data_figS7.nc')
ds_ACC1 = xr.open_dataset(path_input + 'ACC1_bis_data_figS7.nc')
ds_ACC2 = xr.open_dataset(path_input + 'ACC2_data_figS7.nc')
ds_MED = xr.open_dataset(path_input + 'MED_data_figS7.nc')
```

```python
ds_AGULHAS
```

```python
fig, axes = plt.subplots(ncols = 3, nrows = 2, figsize = (10, 5))
ax = axes[0][0]

ds = ds_GS.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')

ax.legend(
    fontsize=9
)
ax.set_title(f'Gulf Stream\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax = axes[0][1]
ds = ds_AGULHAS.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')
ax.set_title('Gulf Stream')
ax.set_title(f'Agulhas\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax.set_xlim([7, 80])
ax.set_ylim([0, 100])



ax = axes[0][2]
ds = ds_CCS.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')
ax.set_title('Gulf Stream')
ax.set_title(f'CCS\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax.set_xlim([7, 80])
ax.set_ylim([0, 100])



ax = axes[1][0]
ds = ds_ACC1.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')
ax.set_title('Gulf Stream')
ax.set_title(f'ACC1\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax.set_xlim([7, 80])
ax.set_ylim([0, 100])

ax = axes[1][1]
ds = ds_ACC2.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')
ax.set_title('Gulf Stream')
ax.set_title(f'ACC2\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax.set_xlim([7, 80])
ax.set_ylim([0, 100])

ax = axes[1][2]
ds = ds_MED.copy()
ax.plot(
    ds.rmax_bin,
    ds.percentage_paired_eddy,
    '-',
    lw=1,
    label='Paired',color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_uncertain_eddy,
    '--',
    lw=1,
    label='Unpaired-d<R$_{max}$', color = 'k',
)
ax.plot(
    ds.rmax_bin,
    ds.percentage_unpaired_eddy,
    ':',
    lw=1,
    label='Unpaired-d>R$_{max}$',color = 'k')
ax2 = ax.twinx()
ax2.bar(ds.rmax_bin, ds.all_eddies, width=2.0, alpha=0.25, color = 'r')
ax.set_title('Gulf Stream')
ax.set_title(f'MED\n(Eddy-Paired + Eddy-Unpaired$_{{Rmax}}$)/Eddy Total = {ds.percentage_of_paired_eddy}', fontsize = 8)
ax.set_xlim([7, 80])
ax.set_ylim([0, 100])

fig.supxlabel('$R_{max}^{SWOT}$ [km]')
fig.supylabel('Occurence [%]')

plt.tight_layout()
ax2.text(100, 400, 'Number of Eddies per R$^{SWOT}_{max}$ bins', color = 'r', fontsize = 14, rotation = 90)

plt.savefig(path_output + 'Fig_S7.png', dpi = 300, bbox_inches = 'tight')
```
