```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
import sys

sys.path.append('..')
from src import compute_pdf
```

```python
path_input = '../data/'
path_output = '../figures/'
```

```python
bin_rmax = np.linspace(0, 80, 30)
fig, axes = plt.subplots(ncols = 3, nrows = 2, sharex = True)
ax = axes[0][0]
ds = xr.open_dataset(path_input + 'GS2_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.plot(bin_center, pdf_swot, color = 'teal', label = 'SWOT')
ax.plot(bin_center, pdf_swot_like, color = 'tomato', label = 'SWOT-like')
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')
ax.set_title('Gulf-Stream')
ax.text(45, np.nanpercentile(pdf_swot, 98), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 90), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.legend(bbox_to_anchor = (0.9, -.1))
ax.set_xticks([15,  30, 45, 60, 75])

ax = axes[0][1]
ds = xr.open_dataset(path_input + 'AGULHAS_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.plot(bin_center, pdf_swot, color = 'teal')
ax.plot(bin_center, pdf_swot_like, color = 'tomato')
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')
ax.set_title('Agulhas')
ax.text(45, np.nanpercentile(pdf_swot, 99), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 96), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.set_xticks([15,  30, 45, 60, 75])

ax = axes[0][2]
ds = xr.open_dataset(path_input + 'CCS_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.plot(bin_center, pdf_swot, color = 'teal')
ax.plot(bin_center, pdf_swot_like, color = 'tomato')
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')
ax.set_title('CCS')

ax.text(45, np.nanpercentile(pdf_swot, 97), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 95), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.set_xticks([15,  30, 45, 60, 75])

ax = axes[1][0]
ds = xr.open_dataset(path_input + 'ACC1_bis_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.plot(bin_center, pdf_swot, color = 'teal')
ax.plot(bin_center, pdf_swot_like, color = 'tomato')
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')
ax.set_title('ACC1')
ax.text(45, np.nanpercentile(pdf_swot_like, 97), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 99), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.set_xticks([15,  30, 45, 60, 75])


ax = axes[1][1]
ds = xr.open_dataset(path_input + 'ACC2_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')

ax.plot(bin_center, pdf_swot, color = 'teal')
ax.plot(bin_center, pdf_swot_like, color = 'tomato')
ax.set_title('ACC2')
ax.text(45, np.nanpercentile(pdf_swot, 98), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 88), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.set_xticks([15,  30, 45, 60, 75])

ax = axes[1][2]
ds = xr.open_dataset(path_input + 'MED_paired_review_test_90perc_binning_eke.nc')
pdf_swot, bin_0center = compute_pdf(ds.rmax_swot, bin_rmax)
pdf_swot_like, bin_center = compute_pdf(ds.rmax_duacs, bin_rmax)
ax.plot(bin_center, pdf_swot, color = 'teal')
ax.plot(bin_center, pdf_swot_like, color = 'tomato')
ax.axvline(np.median(ds.rmax_swot), color = 'teal')
ax.axvline(np.median(ds.rmax_duacs), color = 'tomato')

ax.text(45, np.nanpercentile(pdf_swot, 98), '%.1f km'%np.median(ds.rmax_swot), color = 'teal')
ax.text(45, np.nanpercentile(pdf_swot, 96), '%.1f km'%np.median(ds.rmax_duacs), color = 'tomato')
ax.set_xticks([15,  30, 45, 60, 75])

ax.set_title('MED')
plt.tight_layout()
fig.supxlabel('R$_{max}$ [km]')
fig.supylabel('Occurence [%]')
plt.tight_layout()

plt.savefig(path_output + 'Fig_S6.png', dpi = 300, bbox_inches = 'tight')
```
