```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import sys
sys.path.append('..')
from src import compute_pdf
```

```python
path_in = '../data/'
path_output = '../figures/'
```

```python
ds_ameda_1_8 = xr.open_dataset(path_in + 'eddy_atlas_1_8.nc')
ds_ameda_1_4 = xr.open_dataset(path_in + 'eddy_atlas_1_4.nc')
```

```python
# Eddy radii (km)
r_1_8 = ds_ameda_1_8.r_max.values
r_1_4 = ds_ameda_1_4.r_max.values


bins = np.arange(0, 151, 5)
bin_centers = (bins[:-1] + bins[1:]) / 2

# Number of eddy observations per radius bin
count_1_8 = np.histogram(r_1_8, bins=bins)[0]
count_1_4 = np.histogram(r_1_4, bins=bins)[0]

fig, ax = plt.subplots(figsize=(8, 5))

bin_width = np.diff(bins)[0]
width = 0.4 * bin_width

ax.bar(bin_centers - width/2, count_1_8,
       width=width,
       label='Grid size = 1/8$^{\circ}$')

ax.bar(bin_centers + width/2, count_1_4,
       width=width,
       label='Grid size = 1/4$^{\circ}$')

ax.set_xlabel('$R_{max}$ [km]')
ax.set_ylabel('Number of eddy observations')
ax.legend()
ax.grid(axis='y', alpha=0.3)
ax.set_axisbelow(True)


plt.savefig(path_output + 'Fig_S2.png', dpi = 300, bbox_inches = 'tight')
```
