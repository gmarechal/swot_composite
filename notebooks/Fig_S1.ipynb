```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
```

```python
path_in = '../data/'
path_output = '../figures/'
```

```python

ds_gs = xr.open_dataset(path_in + 'spec_adt_Gulf-Stream.nc')
ds_agulhas = xr.open_dataset(path_in + 'spec_adt_Agulhas.nc')
ds_ccs = xr.open_dataset(path_in + 'spec_adt_CCS.nc')
ds_acc1 = xr.open_dataset(path_in + 'spec_adt_ACC1.nc')
ds_acc2 = xr.open_dataset(path_in + 'spec_adt_ACC2.nc')
ds_med = xr.open_dataset(path_in + 'spec_adt_MED.nc')

```

```python
props = dict(boxstyle='round', facecolor='w')
ds_gs
```

```python
fac = 1
# scale_oi = 20 # in km
fig, axes = plt.subplots(ncols = 6, figsize = (9, 3), sharey = True)
ax = axes[0]
ax.loglog(ds_gs.wavenumber/fac, ds_gs.spec_raw * fac, label = 'L3 product')
ax.loglog(ds_gs.wavenumber/fac, ds_gs.spec_filtered * fac, label = 'Parzen 30 km')
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_gs.spec_filtered * fac), np.amax(ds_gs.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('Gulf Stream')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
fig.legend(bbox_to_anchor=[.25, 1.15])
ax.text(0.82, 0.92, '(a)',
     transform = ax.transAxes, bbox = props, fontsize = 8)


ax = axes[1]
ax.loglog(ds_agulhas.wavenumber/fac, ds_agulhas.spec_raw * fac)
ax.loglog(ds_agulhas.wavenumber/fac, ds_agulhas.spec_filtered * fac)
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_agulhas.spec_filtered * fac), np.amax(ds_agulhas.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('Agulhas')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
ax.text(0.82, 0.92, '(b)',
     transform = ax.transAxes, bbox = props, fontsize = 8)


ax = axes[2]
ax.loglog(ds_ccs.wavenumber/fac, ds_ccs.spec_raw * fac)
ax.loglog(ds_ccs.wavenumber/fac, ds_ccs.spec_filtered * fac)
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_ccs.spec_filtered * fac), np.amax(ds_ccs.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('CCS')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
ax.text(0.82, 0.92, '(c)',
     transform = ax.transAxes, bbox = props, fontsize = 8)


ax = axes[3]
ax.loglog(ds_acc1.wavenumber/fac, ds_acc1.spec_raw * fac)
ax.loglog(ds_acc1.wavenumber/fac, ds_acc1.spec_filtered * fac)
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_acc1.spec_filtered * fac), np.amax(ds_acc1.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('ACC1')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
ax.text(0.82, 0.92, '(d)',
     transform = ax.transAxes, bbox = props, fontsize = 8)

ax = axes[4]
ax.loglog(ds_acc2.wavenumber/fac, ds_acc2.spec_raw * fac)
ax.loglog(ds_acc2.wavenumber/fac, ds_acc2.spec_filtered * fac)
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_acc2.spec_filtered * fac), np.amax(ds_acc2.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('ACC2')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
ax.text(0.82, 0.92, '(e)',
     transform = ax.transAxes, bbox = props, fontsize = 8)


ax = axes[5]
ax.loglog(ds_med.wavenumber/fac, ds_med.spec_raw * fac)
ax.loglog(ds_med.wavenumber/fac, ds_med.spec_filtered * fac)
# ax.plot([1/scale_oi, 1/scale_oi], [10**(-10), 30**1], 'k')
ax.axvline(1/30, color = 'k')
ax.set_ylim([np.amin(ds_med.spec_filtered * fac), np.amax(ds_med.spec_filtered * fac)])
ax.set_xlim([1/500, 1/4])
ax.set_ylim([10**(-13), 30**1])
ax.set_title('MED')
ax.grid(visible=True, which='major', color='k', linestyle='-', alpha = .4)
ax.grid(visible=True, which='minor', color='k', linestyle='--', alpha = .2)
ax.text(0.82, 0.92, '(f)',
     transform = ax.transAxes, bbox = props, fontsize = 8)


fig.supxlabel('Wavenumber [cpkm]')
fig.supylabel('PSD [$m^{2}/cpkm]$')




plt.tight_layout()
plt.savefig(path_output + 'Fig_S1.png', dpi = 300, bbox_inches = 'tight')
```
